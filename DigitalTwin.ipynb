```python
from google.colab import userdata
from groq import Groq

groq_key = userdata.get("Groq")

client = Groq(api_key=groq_key)

MODEL_NAME = "openai/gpt-oss-20b"

print("Groq API connected successfully!")
```

```python
def educational_agent(topic, grade, difficulty, mode):

    if not topic or not topic.strip():
        return "Please enter a topic."

    if mode == "Lesson":

        prompt = f"""
        Generate structured educational lesson notes.

        Topic: {topic}
        Grade: {grade}
        Difficulty: {difficulty}

        Include:
        1. Title
        2. Learning Objectives
        3. Introduction
        4. Detailed Explanation
        5. Important Concepts
        6. Three Worked Examples
        7. Five Practice Questions
        8. Answers with Explanations
        9. Summary

        Use simple language suitable for the selected grade.
        """

    elif mode == "Quiz":

        prompt = f"""
        Generate 10 multiple choice questions.

        Topic: {topic}
        Grade: {grade}
        Difficulty: {difficulty}

        Include four options for each question.
        Give only one correct answer.
        Provide the answer key and explanations at the end.
        """

    else:

        prompt = f"""
        Explain the topic clearly for a student.

        Topic: {topic}
        Grade: {grade}
        Difficulty: {difficulty}

        Include:
        1. Introduction
        2. Detailed Explanation
        3. Important Concepts
        4. Real Life Applications
        5. Examples
        6. Summary
        """

    try:

        response = client.chat.completions.create(

            model=MODEL_NAME,

            messages=[
                {
                    "role": "system",
                    "content": "You are an expert educational AI teacher. Generate accurate, structured, grade-appropriate content."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],

            max_completion_tokens=2500,
            reasoning_effort="low"
        )

        return response.choices[0].message.content

    except Exception as e:

        return "Generation Error: " + str(e)
```

```python
result = educational_agent(
    topic="Trigonometry",
    grade="12",
    difficulty="Hard",
    mode="Lesson"
)

print(result)
```

```python
import gradio as gr

with gr.Blocks(theme=gr.themes.Soft()) as demo:

    gr.Markdown("# 🎓 Educational AI Agent")

    gr.Markdown(
        "Generate Lessons, Quizzes and Explanations using AI"
    )

    with gr.Row():

        topic = gr.Textbox(
            label="Topic",
            placeholder="Enter topic (e.g. Trigonometry)"
        )

        grade = gr.Dropdown(
            choices=[str(i) for i in range(1, 13)],
            value="12",
            label="Grade"
        )

    with gr.Row():

        difficulty = gr.Dropdown(
            choices=["Easy", "Medium", "Hard"],
            value="Medium",
            label="Difficulty Level"
        )

        mode = gr.Radio(
            choices=["Lesson", "Quiz", "Explanation"],
            value="Lesson",
            label="Content Type"
        )

    generate_btn = gr.Button(
        "Generate Content 🚀",
        variant="primary"
    )

    output = gr.Markdown(
        label="Generated Output"
    )

    generate_btn.click(
        fn=educational_agent,
        inputs=[topic, grade, difficulty, mode],
        outputs=output
    )

demo.queue().launch(share=True, debug=True)
```
